# **Transformation**

> **Fix (Task 3/4 review):** the transformation steps originally ran on `df` (the *raw* data,
> still containing NaNs, duplicates, 999999 / negative salaries and rows with missing `Attrition`).
> They now run on a copy of the cleaned data `df_clean`, stored as `df_transformed`, so the
> cleaned dataset itself stays untouched for EDA.


Normalize Numerical Features


---



```python
# FIX: work on a copy of the CLEANED data, not the raw df
df_transformed = df_clean.copy()
df_transformed.dtypes
```

```python
numerical_cols = [
    'Age',
    'TenureYears',
    'PerformanceRating',
    'MonthlySalary'
]
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
df_transformed[numerical_cols] = scaler.fit_transform(df_transformed[numerical_cols])
```

```python
df_transformed[numerical_cols].head()
```

Encode Categorical Variables


---



```python
categorical_cols = [
    'Gender',
    'JobRole'
]
df_transformed = pd.get_dummies(
    df_transformed,
    columns=categorical_cols,
    drop_first=True
)
```

```python
df_transformed.head()
```

```python
df_transformed.dtypes
```

```python
df_transformed['Attrition'] = df_transformed['Attrition'].map({
    'Yes': 1,
    'No': 0
})
df_transformed['Attrition'].value_counts()
```

Generate Interaction Features


---



```python
df_transformed['Age_Performance_Interaction'] = (
    df_transformed['Age'] * df_transformed['PerformanceRating']
)
```

```python
df_transformed['Tenure_Performance_Interaction'] = (
    df_transformed['TenureYears'] * df_transformed['PerformanceRating']
)
```

```python
df_transformed['Salary_Performance_Interaction'] = (
    df_transformed['MonthlySalary'] * df_transformed['PerformanceRating']
)
```

```python
df_transformed.head()
```

```python
df_transformed.columns
```

```python
df_transformed.shape
```

Additional Engineered Features (required by Milestone 1)

---


```python
# Salary-to-performance ratio and tenure groups use the ORIGINAL (unscaled) values from df_clean
df_transformed['Salary_to_Performance_Ratio'] = (
    df_clean['MonthlySalary'] / df_clean['PerformanceRating']
)
df_transformed['Salary_to_Performance_Ratio'] = scaler.fit_transform(
    df_transformed[['Salary_to_Performance_Ratio']]
)

tenure_groups = pd.cut(
    df_clean['TenureYears'],
    bins=[-np.inf, 2, 5, 10, np.inf],
    labels=['0-2', '3-5', '6-10', '10plus']
)
df_transformed = df_transformed.join(
    pd.get_dummies(tenure_groups, prefix='TenureGroup', drop_first=True)
)
df_transformed.head()
```

Validate & Save Transformed Dataset

---


```python
print('Shape:', df_transformed.shape)
print('Missing values:', df_transformed.isnull().sum().sum())
print('Duplicates:', df_transformed.duplicated().sum())
print(df_transformed['Attrition'].value_counts())

df_transformed.to_csv('employee_attrition_transformed.csv', index=False)
```
